## 3. nashpy — equilibrios de Nash

````python
import nashpy as nash
print(f"nashpy {nash.__version__}")

# Test rápido: Dilema del Prisionero
A = np.array([[-1, -3], [0, -2]])  # pagos Jugador 1
B = np.array([[-1,  0], [-3, -2]]) # pagos Jugador 2
pris = nash.Game(A, B)
eqs = list(pris.support_enumeration())
print(f"Equilibrios encontrados: {len(eqs)}")
print(f"Equilibrio: {eqs[0]}")
print("✓ nashpy OK — equilibrio del Dilema del Prisionero calculado")
````

## 4. axelrod — torneo del Dilema Iterado

````python
import axelrod
print(f"axelrod {axelrod.__version__}")

# Mini torneo de prueba
players = [
    axelrod.TitForTat(),
    axelrod.Defector(),
    axelrod.Cooperator(),
]
tournament = axelrod.Tournament(players, turns=20, repetitions=3, seed=42)
results = tournament.play(progress_bar=False)
print(f"Ranking: {[str(p) for p in results.ranked_names]}")
print("✓ axelrod OK — torneo ejecutado")
````
