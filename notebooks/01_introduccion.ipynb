## 1. Un juego en forma normal

Un juego G = (N, S, u) se define por:
- **N** = {1, ..., n} jugadores
- **Sᵢ** = estrategias disponibles para el jugador i
- **uᵢ: S₁ × ... × Sₙ → ℝ** función de pago

Con 2 jugadores y estrategias discretas, lo representamos con dos matrices A (pagos J1) y B (pagos J2).

```python
# Dilema del Prisionero
# Filas: J1 (Cooperar=0, Traicionar=1)
# Columnas: J2 (Cooperar=0, Traicionar=1)

A = np.array([[3, 0],   # pagos de J1
              [5, 1]])

B = np.array([[3, 5],   # pagos de J2
              [0, 1]])

strategies = ['Cooperar', 'Traicionar']

print('Matriz de pagos — Dilema del Prisionero')
print(f'{"":12}', '  '.join(f'J2={s}' for s in strategies))
for i, s1 in enumerate(strategies):
    row = '  '.join(f'({A[i,j]}, {B[i,j]})' for j in range(len(strategies)))
    print(f'J1={s1:10} {row}')
```

## 2. Acceder a pagos de un perfil

Dado un perfil de estrategias (s₁, s₂), los pagos son A[s₁, s₂] y B[s₁, s₂].

```python
def get_payoffs(A, B, s1, s2, strats):
    """Devuelve los pagos del perfil (strats[s1], strats[s2])."""
    return A[s1, s2], B[s1, s2]

# Perfil: J1 coopera (0), J2 traiciona (1)
u1, u2 = get_payoffs(A, B, 0, 1, strategies)
print(f'Perfil (Cooperar, Traicionar): u1={u1}, u2={u2}')

# Iterar todos los perfiles
print('\nTodos los perfiles:')
for s1, s2 in itertools.product(range(2), repeat=2):
    u1, u2 = get_payoffs(A, B, s1, s2, strategies)
    print(f'  ({strategies[s1][:3]}, {strategies[s2][:3]}): u1={u1}, u2={u2}')
```

## 3. Visualización de la matriz

Un heatmap de los pagos de J1 para identificar visualmente las estrategias más rentables.

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(9, 3.5))

for ax, M, title, cmap in [(ax1, A, 'Pagos J1', 'Blues'), (ax2, B, 'Pagos J2', 'Oranges')]:
    im = ax.imshow(M, cmap=cmap, vmin=0, vmax=5)
    ax.set_xticks(range(2)); ax.set_yticks(range(2))
    ax.set_xticklabels(strategies); ax.set_yticklabels(strategies)
    ax.set_xlabel('Estrategia J2'); ax.set_ylabel('Estrategia J1')
    ax.set_title(title)
    for i in range(2):
        for j in range(2):
            ax.text(j, i, str(M[i, j]), ha='center', va='center', fontsize=16, fontweight='bold')
    plt.colorbar(im, ax=ax)

plt.tight_layout()
plt.suptitle('Dilema del Prisionero', y=1.02, fontsize=13, fontweight='bold')
plt.show()
```

## 4. Decisión ordinaria vs. estratégica

La decisión es **estratégica** cuando el resultado depende de las acciones de más de un agente. Formalmente: si ∃ i tal que uᵢ(sᵢ, s₋ᵢ) ≠ uᵢ(sᵢ, s'₋ᵢ) para algún s'₋ᵢ ≠ s₋ᵢ, el juego es no trivial.

```python
def is_strategic(A):
    """True si los pagos de J1 dependen de la estrategia de J2."""
    # Si todas las columnas de A son iguales, J1 puede ignorar a J2
    return not np.all(A == A[:, [0]])  # compara cada columna con la primera

# Caso no estratégico: J1 siempre obtiene lo mismo sea cual sea la decisión de J2
A_trivial = np.array([[3, 3], [5, 5]])
print('A trivial (misma columna):', is_strategic(A_trivial))  # False
print('A del Dilema del Prisionero:', is_strategic(A))        # True
```
